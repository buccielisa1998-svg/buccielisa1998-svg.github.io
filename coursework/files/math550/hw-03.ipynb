a) We fit the model for starting salary (in \$thousands):

$$
\widehat{Y}=50+20X_1+0.07X_2+35X_3+0.01(X_1\cdot X_2)-10(X_1\cdot X_3),
$$

where

$$
X_1=\mathrm{GPA},\qquad X_2=\mathrm{IQ},\qquad X_3=\mathrm{Level}\ (1=\mathrm{College},\ 0=\mathrm{High\ School}).
$$

For a fixed GPA and IQ, we compare the predicted salary for college graduates ($X_3=1$) and high school graduates ($X_3=0$).

$$
\widehat{Y}_{\mathrm{College}}=50+20\,\mathrm{GPA}+0.07\,\mathrm{IQ}+35(1)+0.01(\mathrm{GPA}\cdot\mathrm{IQ})-10(\mathrm{GPA}\cdot1),
$$

$$
\widehat{Y}_{\mathrm{HS}}=50+20\,\mathrm{GPA}+0.07\,\mathrm{IQ}+35(0)+0.01(\mathrm{GPA}\cdot\mathrm{IQ})-10(\mathrm{GPA}\cdot0).
$$

Subtracting:

$$
\widehat{Y}_{\mathrm{College}}-\widehat{Y}_{\mathrm{HS}}=35-10\cdot\mathrm{GPA}.
$$

$$
\begin{cases}
\text{If GPA}<3.5, & \widehat{Y}_{\mathrm{College}}>\widehat{Y}_{\mathrm{HS}} \quad\Rightarrow\quad \text{College earns more},\\
\text{If GPA}=3.5, & \widehat{Y}_{\mathrm{College}}=\widehat{Y}_{\mathrm{HS}},\\
\text{If GPA}>3.5, & \widehat{Y}_{\mathrm{College}}<\widehat{Y}_{\mathrm{HS}} \quad\Rightarrow\quad \text{High School earns more}.
\end{cases}
$$

Therefore, the correct statement is:

> (iii) For a fixed IQ and GPA, high school graduates earn more, on average, than college graduates provided that the GPA is high enough.


b) We use the fitted model

$$
\widehat{Y}=50+20X_1+0.07X_2+35X_3+0.01(X_1\cdot X_2)-10(X_1\cdot X_3),
$$

where $X_1=\mathrm{GPA}$, $X_2=\mathrm{IQ}$, and $X_3=\mathrm{Level}\ (1=\mathrm{College},\ 0=\mathrm{High\ School})$.

For a college graduate with $\mathrm{GPA}=4.0$ and $\mathrm{IQ}=110$, we set $X_1=4.0$, $X_2=110$, $X_3=1$. Then

$$
\widehat{Y}=50+20(4.0)+0.07(110)+35(1)+0.01(4.0\cdot110)-10(4.0\cdot1).
$$

Simplifying,

$$
\widehat{Y}=50+80+7.7+35+4.4-40,
$$

$$
\widehat{Y}=137.1.
$$

Therefore, the predicted starting salary is

$$
137.1\ \text{thousand dollars}\qquad \approx \$137{,}100.
$$

c) A small interaction coefficient does not imply a little evidence of an interaction effect. If we consider from the model

$$
\widehat{Y}=\cdots+0.07\,\mathrm{IQ}+0.01(\mathrm{GPA}\cdot\mathrm{IQ})+\cdots,
$$

the marginal effect of IQ depends on GPA:

$$
\frac{\partial\widehat{Y}}{\partial\mathrm{IQ}}=0.07+0.01\,\mathrm{GPA}.
$$

Thus, increasing GPA by 1 point increases the effect of IQ on salary by 0.01 (in thousands), i.e. \$10 per IQ point. Over realistic ranges, like GPA $=2$ vs $4$ and IQ range of 30 points, this can amount to several hundred dollars, which it’s not negligible.

a) We collect $n=100$ observations from a dataset with a single predictor $X$ and quantitative response $Y$. We fit both a linear model

$$
Y=\beta_0+\beta_1X+\varepsilon,
$$

and a cubic model

$$
Y=\beta_0+\beta_1X+\beta_2X^2+\beta_3X^3+\varepsilon.
$$

### a)Suppose the true relationship is linear:

$$
Y=\beta_0+\beta_1X+\varepsilon.
$$

The cubic model class contains the linear model as a special case (by setting $\beta_2=\beta_3=0$). Because ordinary least squares chooses coefficients to minimize the training residual sum of squares (RSS) within a model class, minimizing over a larger class cannot increase RSS. Therefore,

$$
\mathrm{RSS}_{\mathrm{train}}(\mathrm{cubic})\leq\mathrm{RSS}_{\mathrm{train}}(\mathrm{linear}).
$$

With noisy data, the cubic model will often achieve a strictly smaller training RSS, although in some samples the extra terms may shrink towards zero, yielding nearly equal RSS.

b) For test error, we consider the bias–variance decomposition:

$$
\mathrm{E}\left[(Y_0-\widehat{f}(X_0))^2\right]=\sigma^2+\mathrm{Bias}^2[\widehat{f}(X_0)]+\mathrm{Var}[\widehat{f}(X_0)].
$$

Both linear and cubic models have zero bias when the true relationship is linear. However, the cubic model estimates extra unnecessary parameters, which increases the variance of predictions. Therefore the expected test RSS is higher for the cubic model:

$$
\mathrm{E}[\mathrm{RSS}_{\mathrm{test}}(\mathrm{linear})]<\mathrm{E}[\mathrm{RSS}_{\mathrm{test}}(\mathrm{cubic})].
$$

c) Now suppose the true relationship between $X$ and $Y$ is not linear. Again, the cubic model class contains the linear one. Thus the training RSS for the cubic model will be less than or equal to that of the linear model:

$$
\mathrm{RSS}_{\mathrm{train}}(\mathrm{cubic})\leq\mathrm{RSS}_{\mathrm{train}}(\mathrm{linear}).
$$

Typically it is strictly smaller, since the extra terms allow the cubic model to capture curvature in the data (and even noise).
